```python
import pandas as pd

df_whole = pd.DataFrame({
    "Search Method": ["Random Search", "Genetic Algorithm", "Annealing"],
    "Mean": [1.23, 4.56, 7.89],
    "Median": [0.12, 0.34, 0.56],
    "IQR": [1.0, 1.0, 1.0],
    "min. rate": [1.0, 1.0, 1.0],
    "max rate": [1.0, 1.0, 1.0],
    "identical samples": [1.0, 1.0, 1.0]
})

df_topk = pd.DataFrame({
    "Search Method": ["Random Search", "Genetic Algorithm", "Annealing"],
    "Mean": [0.95, 3.21, 6.54],
    "Median": [0.10, 0.30, 0.50],
    "IQR": [0.8, 0.9, 1.1],
})

# combine side-by-side (drop duplicate column)
df_combined = pd.concat(
    [df_whole, df_topk.drop(columns=["Search Method"])],
    axis=1
)

display(df_combined)
```

```python
latex = (
    df_combined.style
    .hide(axis="index")
    .format("{:.2f}", subset=df_combined.columns[1:])
    .to_latex(
        hrules=True,
        column_format="l|cccccc|" + "ccc",  # 1 + 5 + 5 columns
        multicol_align="c"
    )
)

header = r"""
\multicolumn{1}{c}{} &
\multicolumn{6}{c}{Whole distribution} &
\multicolumn{3}{c}{Top\_k solutions} \\
"""

latex = latex.replace(r"\toprule", r"\toprule" + header)
```
